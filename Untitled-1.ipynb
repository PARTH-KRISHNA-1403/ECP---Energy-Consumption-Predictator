```python
import pandas as pd
import sklearn as sc
from matplotlib import pyplot as plt
```

```python
print("Loading The Dataset ...")
temp_df= pd.read_csv("dataset.csv")
```

```python
df= temp_df[
    (~temp_df["State"].isin(["India Total","Other"])) &
    (temp_df["Category"]=="Electricity generation") &
    (temp_df["Variable"]=="Total Generation") &
    (temp_df["Unit"]=="GWh")].copy()
df["Date"]=pd.to_datetime(df["Date"],format="%d-%m-%Y") 
df["Month"]=df["Date"].dt.month
df["Year"]=df["Date"].dt.year
df = df.sort_values(["State","Date"])

```

```python
def get_season(month):
    if (month in [12,1,2]):
        return "Winter"
    elif (month in [3,4,5]):
        return "Summer"
    elif (month in [6,7,8,9]):
        return "Monsoon"
    else:
        return "Autumn"

df["Season"]= df["Month"].apply(get_season)
```

```python

df["Lag1"]= df.groupby("State")["Value"].shift(1)
df["Lag12"]= df.groupby("State")["Value"].shift(12)

ml_data = df.dropna(subset=["Lag1","Lag12","Value"]).copy()
ml_data = pd.get_dummies(ml_data,drop_first=True,columns=["Season","State"])
exclude_cols= ['Country', 'State type', 'Date', 'Category', 'Subcategory', 'Variable',
       'Unit', 'Value']
features = [col for col in ml_data.columns if col not in exclude_cols]

X= ml_data[features]
Y= ml_data['Value']
```
